```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(12, 6))

plt.plot(df["Name"], df["Total"], marker="o")

plt.xlabel("Student Name")
plt.ylabel("Total Marks")
plt.title("Total Marks of Students")

plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(df["Maths"], df["Physics"])

plt.xlabel("Maths Marks")
plt.ylabel("Physics Marks")
plt.title("Maths vs Physics Marks")

plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 6))

plt.hist(df["Total"], bins=5, edgecolor="black")

plt.xlabel("Total Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Total Marks")

plt.show()
```

```python
gender_count = df["Gender"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    gender_count,
    labels=gender_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Gender Distribution")

plt.show()
```

```python
subjects = ["Maths", "Physics", "Chemistry", "English", "Computer"]

average_marks = df[subjects].mean()

plt.figure(figsize=(9, 6))

plt.bar(average_marks.index, average_marks.values)

plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.title("Average Marks by Subject")

plt.show()
```

```python
top10 = df.sort_values("Total", ascending=False).head(10)

plt.figure(figsize=(10, 6))

plt.barh(top10["Name"], top10["Total"])

plt.xlabel("Total Marks")
plt.ylabel("Student Name")
plt.title("Top 10 Students by Total Marks")

plt.gca().invert_yaxis()

plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.bar(df["Name"], df["Maths"], label="Maths")
plt.bar(df["Name"], df["Physics"], bottom=df["Maths"], label="Physics")

plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.title("Maths and Physics Marks")

plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()

plt.show()
```
